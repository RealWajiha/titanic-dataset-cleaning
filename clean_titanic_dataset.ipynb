# Task 01: Clean & Prepare a Dataset
## Dataset: Titanic Passenger Dataset

**Objective:** To clean and prepare a real-world dataset for analysis and machine learning.

### Tasks
- Load the dataset
- Identify and handle missing values
- Remove duplicate records
- Correct inconsistent values
- Convert columns to appropriate data types
- Save the cleaned dataset

```python
import pandas as pd
import numpy as np
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nStatistical Summary:")
df.describe(include="all").T
```

CHECK MISSING VALUES

```python
# Check missing values before cleaning
print("Missing values before cleaning:")
print(df.isnull().sum())

# Fill missing numerical values with median
if "Age" in df.columns:
    df["Age"] = df["Age"].fillna(df["Age"].median())

if "Fare" in df.columns:
    df["Fare"] = df["Fare"].fillna(df["Fare"].median())

# Fill missing categorical values with mode
for col in ["Embarked", "Sex"]:
    if col in df.columns and df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

# Cabin has many missing values, so drop it
if "Cabin" in df.columns:
    df = df.drop(columns=["Cabin"])

print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

BASELINE METRICS CODE

```python
# Record baseline metrics before cleaning
df_before = df.copy()

before_metrics = {
    "Rows": df.shape[0],
    "Columns": df.shape[1],
    "Duplicate Rows": int(df.duplicated().sum()),
    "Missing Cells": int(df.isnull().sum().sum()),
    "Integer Columns": df.select_dtypes(include="integer").shape[1],
    "Float Columns": df.select_dtypes(include="float").shape[1],
    "Object Columns": df.select_dtypes(include="object").shape[1]
}

print("Baseline quality metrics recorded successfully!")
print(before_metrics)
```

REMOVE DUPLICATE VALUE

```python
print("Duplicates before cleaning:", df.duplicated().sum())

df = df.drop_duplicates().reset_index(drop=True)

print("Duplicates after cleaning:", df.duplicated().sum())
print("Dataset shape:", df.shape)
```

CHECK INCORRECT AND INCONSISTANT VALUE

```python
# Check unique categorical values
for col in df.select_dtypes(include="object").columns:
    print(f"\n{col}:")
    print(df[col].value_counts(dropna=False).head(20))

# Check impossible or invalid numerical values
print("\nInvalid Age values:", (df["Age"] < 0).sum())
print("Invalid Fare values:", (df["Fare"] < 0).sum())
print("Invalid SibSp values:", (df["SibSp"] < 0).sum())
print("Invalid Parch values:", (df["Parch"] < 0).sum())
```

CORRECT DATATYPES

```python
# Convert columns to appropriate data types
df["PassengerId"] = df["PassengerId"].astype("int64")
df["Survived"] = df["Survived"].astype("int64")
df["Pclass"] = df["Pclass"].astype("int64")
df["SibSp"] = df["SibSp"].astype("int64")
df["Parch"] = df["Parch"].astype("int64")

df["Age"] = df["Age"].astype("float64")
df["Fare"] = df["Fare"].astype("float64")

print("Updated data types:")
df.dtypes
```

FINAL QUALITY CHECK

```python
print("Final Dataset Shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nFinal data types:")
print(df.dtypes)

df.head(10)
```

SAVE CLEAN DATASET

```python
output_file = "titanic_cleaned.csv"

df.to_csv(output_file, index=False)

print(f"Cleaned dataset saved successfully as: {output_file}")
```
